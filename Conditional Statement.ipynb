```python
#Nested if else 
age=int(input("Enter Age: "))
idPass=False
if(age>=18):
    if(idPass == True):
        print("Student Entry Allowed")
    else:
        print("Student have no ID. Entry Not Allowed")
else:
    print("Student age is not Valid. Entry Not Allowed")
```

```python
pin=int(input("Enter Pin"))
balance =25000

if(pin == 1234):
    amount=int(input("Enter Amount you want to Withdraw")) #1000
    if(amount%100 == 0):
        if(balance>=amount):
            balance -= amount
            print("Remaining Balance is: ",balance)
        else:
            print("Paisa Nhi Hai")

    else:
        print("Enter Valid Amount")

else:
    print("Invlaid Pin")
```

```python
java=int(input("Enter Marks of Java"))
if(java>40):
    python=int(input("Enter Marks of Python"))
    if(python>40):
        dbms=int(input("Enter Marks of DBMS"))
        if(dbms>40):
            total_marks=java+python+dbms
            per=total_marks/3
            print("Total Marks: ",total_marks)
            print("Student Per: ",per)
            if(per>=75):
                print("A+")
            elif(per>=60):
                print("First Division")
            elif(per>=50):
                print("Second Division")
            else:
                print("Pass")
        else:
            print("Fail in DBMS")
    else:
        print("Fail in Python")
else:
    print("Fail in Java")
```

```python
# ATM Withdrawal
balance=45000
pin=int(input("Enter Your Pin"))
if(pin == 1234):
    amount=int(input("what amount do you want to withdraw"))
    if(amount%100 == 0):
        if(amount <= balance):
            balance -= amount 
            print("Withdraw amount is: ",amount)
            print("Remaing balance is: ",balance)
        else:
            print("Paisa Nahi hai")
    else:
        print("Invalid amount")
else:
    print("Incorrect Pin")
```

```python
#Bank Loan Eligibility
age=int(input("Enter Your Age:"))
if(age>=21):
    salary=int(input("Enter Your Salary"))
    if(salary>=25000):
        creditScore=int(input("Enter Your Credit Score: "))
        if(creditScore>=750):
            print("Loan Approved");
            print("Interest Rate: 8%");
        elif(creditScore >= 650):
            print("Loan Approved");
            print("Interest Rate: 10%");
        else:
            print("Loan Rejected")
    else:
        print("Salary is too Low")
else:
    print("Age is not eligible")
```

```python
#Login + Role Based Access
userName=input("Enter UserName: ")
if(userName == "Admin"):
    password=input("Enter Password: ")
    if(password == "12345"):
        role=input("Enter Role: ")
        if(role == "Admin"):
            print("You Have Full Access")
        elif(role == "Trainer"):
            print("Trainer Dashboard")
        elif(role == "Student"):
            print("Student Dashboard")
        else:
            print("Invalid Role")
    else:
        print("Invalid Password")
else:
    print("Invalid User Name")
```
