```python
numbers = [10, -5, 20, -8, 30, -2]
result=[ex if ex>0 else 0 for ex in numbers]
print(result)
```

```python
transactions = [5000, -2000, 3000, -1000, 7000, -500]
deposits=[ex for ex in transactions if ex>0]
print(deposits)
```

```python
prices = [500, 1200, 800, 2000, 1500]
result=[ex*0.90 if ex>=1000 else ex*0.95 for ex in prices]
print(result)
```

```python
employees = [
{"name": "Amit", "salary": 45000},
{"name": "Rahul", "salary": 60000},
{"name": "Neha", "salary": 75000},
{"name": "Rohit", "salary": 35000}
]

result=[ex['name' ] for ex in employees if ex['salary']>50000]
print(result)
```
