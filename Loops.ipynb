```python
data=[22,32,45,-2,87,-1,89,-4]
posCount=0
negCount=0
i=0
while(i<len(data)):
    if data[i]>0:
        posCount+=1
    else:
        negCount+=1
    i+=1
print(posCount)
print(negCount)
```

```python
data="Python Programming"
count=0
i=0
while(i<len(data)):
    if data[i] == "P" or data[i] == 'p':
        count+=1
    i+=1
print(count)
```
